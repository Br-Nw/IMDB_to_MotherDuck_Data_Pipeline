```python
import duckdb
import pandas as pd
import numpy as np
from dotenv import load_dotenv
import os

load_dotenv()
motherduck_token = os.getenv("MOTHERDUCKTOKEN")

with duckdb.connect(f'md:?motherduck_token={motherduck_token}') as con:
    raw_movie_df = con.sql("SELECT * FROM imdb_analytics.bronze.raw_movie_data;").df()
```

```python
df = pd.DataFrame.from_dict(list(raw_movie_df["result"]))
df.head(3)
```

```python
for column in ["Series_Title", "Certificate", "Genre"]:
    print(df[column].value_counts())
    print("\n")
```
